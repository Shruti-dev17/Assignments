## Q1- 
   ### a.
       :To create a simple line plot we use plot() function from matplotlib.pyplot.
       :we put data of x axis , y axis along with linestyle and color as parameters fot plot() function to create a simple line plot.
   ### b.
       :to customize the plot by adding title ,axis labels and a grid we use plt.title(),plt.xlabel(),plt.ylabel(),plt.grid() respectively.

```python
x=[1,2,3,4,5]
y=[10,15,25,30,50]

plt.plot(x,y,linestyle='-',color='r')
plt.title("Line Plot0")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid()
plt.show()
```

## Q2
  ### a.
       :To create a bar graph we use bar()function.
        and add data od x , y as main parameters.
  ### b.
       :To customize colors we add color parameter in bar() and give it a list of color that we want.
       :we use title() function.

```python

students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]

plt.bar(students,marks,color=['r','b','y','g'])
plt.title("Students-marks Bar Graph")
plt.show()
```

## Q3
  ### a.
       :To create a pie chart we use pie() function and put the value along with its label as parameter.We also add colors and autopct as parameter to make the pie understandable.

  ### b.
       :To highlight the region with highest revenue -
          Step 1- Identify the index of the region with highest revenue.
          Step 2 - Use explode()function 
          Step 3 - As parameter put a list where only the index of the region with highest revenue is non zero.

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

plt.pie(revenue,labels=regions,colors=['r','b','y','g'],autopct='%1.1f%%',explode=[0.1,0,0,0])
plt.show()
```

## Q4
     :To generate a histogram to show the frequency distribution of randomly generated integers between 1 and 100 -:
   ### Step 1
         :import numpy to generate random integers 
         :use random.randint() to generate 1000 values in between 1 and 100 at random.
         :use hist()function to create a histogram
         :put the 'sample'(generated values) as the parameter along with color,edgecolor and bins(number of boxes) as needed.

```python
import numpy as np
sample= np.random.randint(1,100,1000)
plt.hist(sample,color='y',bins=50,edgecolor='black')
plt.show()
```
