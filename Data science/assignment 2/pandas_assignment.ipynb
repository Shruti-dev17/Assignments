```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Q1-
    :create a DataFrame with given dictionery using pd.DataFrame().
    :I also used subplots(),bar() funcions from matplot.pyplot(plt) library to add graphs for visualisation as i was told to do in feedback.
    :set_title() is used for title of axes.
    :set_xlabel(),set_ylabel() are used for giving heading to x axis and y axis respectively.
    :plt.show() is used to show the graph.

```python
data=pd.DataFrame({
      'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
      'Age': [24, 27, 22, 32, 29],
      'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
      'Salary': [45000, 54000, 50000, 62000, 47000]
  })

fig,axes= plt.subplots(3,1, figsize=(8,15))
ax=axes[0]
ax.bar(data['Name'],data['Age'],color=['r','b','y','g','purple'])
ax.set_title("Age of Employees") 
ax.set_xlabel("Name of Employees")
ax.set_ylabel("Age")
ax1=axes[1]
ax1.bar(data['Name'],data['Department'],color=['r','b','y','g','purple'])
ax1.set_title("Department of Employees") 
ax1.set_xlabel("Name of Employees")
ax1.set_ylabel("Deapartment")
ax2=axes[2]
ax2.bar(data['Name'],data['Salary'],color=['r','b','y','g','purple'])
ax2.set_title("Salary of Employees") 
ax2.set_xlabel("Name of Employees")
ax2.set_ylabel("Salary")
plt.show()
```

    ### Step3:

```python
print(data.head(),"\n")

print("summary statistics:\n",data[['Age','Salary']].describe(),"\n")

Average=data.groupby('Department')['Salary'].mean()
print("Average Salary of employees in 'HR'department:\n",Average[['HR']],"\n")


```

## Q4 -

  ### Step 1:

   ### Step2:
