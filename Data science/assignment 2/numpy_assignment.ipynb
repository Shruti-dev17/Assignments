## Q1-
### step 1-
           :create an array with elements 1-20 using arange() function in numpy. 
### step 2- 
           :to find sum of elements in array = use sum() function 
           :to find mean of elements in array= use mean() function 
           :to find median of elements in array= use median() function 
           :to findstandard deviation of elements in array = use std() function 
### step 3 -
           :to find the index of each element greater than 10 we use where() function in numpy this function helps in selection of elements and gives index of the elements as result.

```python
arr= np.arange(1,21)

print("Sum of elements :",np.sum(arr),"\n")
print("Mean of elements:",np.mean(arr),"\n")
print("Median of elements:", np.median(arr),"\n")
print( "Standard deviation of elements:", np.std(arr),"\n" )

indices = np.where(arr>10)
print( "index of elements greater than 10: ",indices)


```

## Q2-
### Step 1 - 
             :create a array with elements between 1-16 using arange() function 
             :Shape it into 4x4 matrix using reshape() function
### Step 2- 
             :use print() function to print the array
             :use transpose() function in numpy to get the transpose of matrix
             :use sum() and axis to get row-wise and column-wise sum of elements.
                   - axis =1: for row-wise sum 
                   - axis =0: for column-wise sum

```python
arr2= np.arange(1,17).reshape(4,4)

print("4x4 array :\n",arr2,"\n")
print("transpose of array:\n",np.transpose(arr2),"\n")
print("row-wise sum of array:",np.sum(arr2,axis=1),"\n")
print("column-wise sum of array:",np.sum(arr2,axis=0))
```

## Q3- 
### Step 1-
           :random.randint() function in numpy is used to fill random integers as elements in array.
           it also helps in shaping the array into 2D.
### Step 2-
           :to get element- wise addition,subtraction and multiplication we use (+,-,*) operators respectively.
### Step 3-
           :to find dot product of 2 array we use dot() function from numpy.
     

```python
matrix1 = np.random.randint(1,21,(3,3))
matrix2 = np.random.randint(1,21,(3,3))

print("element-wise addition:\n", matrix1 + matrix2)
print("element-wise subtraction:\n",matrix1 - matrix2)
print("element-wise multiplication:\n",matrix1 * matrix2)
print("Dot product:\n",np.dot(matrix1,matrix2))
```

## Q4-
### Step 1-
           :create a 1D array with 12 elements using arange()function.
### Step2-
           :reshape it into 2D array of 3x4 by using reshape()function.
### Step3- 
           :Slice the first 2 rows [0:2] and the last 2 columns[2:] from the 2D array with the slicing syntax 
                    -array_name[row,column]


```python
arr4= np.arange(12)
print("1D array:\n",arr4)
reshaped=arr4.reshape(3,4)
print("3x4 2D array:\n",reshaped)
print("Slicing first 2 rows and last 2 columns:\n",reshaped[0:2,2:])
```
